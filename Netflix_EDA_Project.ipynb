# Netflix Movies and TV Shows - EDA Project

### Objective:
To analyze Netflix's content library and find insights about Movies vs TV Shows, Top Countries, Genres and Growth over years.

**Dataset:** netflix_titles.csv (From Kaggle - 8807 rows, 12 columns)

**Tools:** Python, Pandas, Matplotlib, Seaborn

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10,5)

df = pd.read_csv('netflix_titles.csv')
df.head()
```

## 1. Data Understanding and Cleaning
Checking null values and handling missing data.

```python
print(f"Dataset Shape: {df.shape}")
display(df.isnull().sum())
df['director'].fillna('Not Given', inplace=True)
df['country'].fillna('Not Given', inplace=True)
df['cast'].fillna('Not Given', inplace=True)
print("Data Cleaned Successfully")
```

## 2. Movies vs TV Shows Distribution

```python
sns.countplot(x='type', data=df, palette='Set2')
plt.title('Distribution of Movies vs TV Shows on Netflix', fontsize=14, fontweight='bold')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()

print(df['type'].value_counts())
```

## 3. Top 10 Countries Producing Content

```python
df['country'].value_counts().head(10).plot(kind='bar', color='coral')
plt.title('Top 10 Countries on Netflix', fontsize=14, fontweight='bold')
plt.xticks(rotation=45)
plt.show()
```

## 4. Content Growth Over the Years

```python
year_df = df['release_year'].value_counts().sort_index().tail(20)
year_df.plot(kind='line', marker='o', linewidth=2, color='red')
plt.title('Netflix Content Growth (2005-2021)', fontsize=14, fontweight='bold')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()
```

## 5. Conclusion and Insights
- **Movies (70%)** are more than TV Shows (30%) on Netflix.
- **United States** is the top producer of content.
- After **2015**, Netflix content growth increased exponentially.
- Most common rating is **TV-MA** and most common genre is **International Movies**.
